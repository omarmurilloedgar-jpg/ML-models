```python
from sklearn.datasets import make_regression
import pandas as pd

# 28 candidate IM names used in Ding & Feng
feature_names = [
    "PGA",
    "SMA",
    "CAV",
    "Arms",
    "Ia",
    "Ic",
    "EDA",
    "PGV",
    "SMV",
    "CAD",
    "Vrms",
    "SED",
    "FI",
    "PGD",
    "Drms",
    "SD",
    "Varatio",
    "ASI",
    "EPA",
    "Sa_T1",
    "Sa_0.2s",
    "Sa_1.0s",
    "Sa_2.0s",
    "Sa_avg",
    "VSI",
    "HI",
    "Sv_T1",
    "Sd_T1"
]

X_temp, y_temp = make_regression(
    n_samples=320,
    n_features=28,
    n_informative=10,
    noise=10,
    random_state=42
)

df = pd.DataFrame(X_temp, columns=feature_names)
df["Max_IDR"] = y_temp

df.head()
```

Create train-test split with Sklearn

```python
from sklearn.model_selection import train_test_split

feature_names = df_clean.drop(columns=["Max_IDR"]).columns.tolist()

X = df_clean.drop(columns=["Max_IDR"])
y = df_clean["Max_IDR"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Features:", feature_names)
```

Standardize Earthquake Intensity Measures

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train_pd = pd.DataFrame(
    X_train_scaled,
    columns=X.columns,
    index=X_train.index
)

X_test_pd = pd.DataFrame(
    X_test_scaled,
    columns=X.columns,
    index=X_test.index
)

print(X_train_pd.head())
```

## **Linear Regressions**

Linear Regression

```python
# Linear regression baseline model
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

baseline = LinearRegression()

baseline.fit(X_train_scaled, y_train)

baseline_preds = baseline.predict(X_test_scaled)

train_r2 = baseline.score(X_train_scaled, y_train)
mse = mean_squared_error(y_test, baseline_preds)
rmse = np.sqrt(mse)
test_r2 = r2_score(y_test, baseline_preds)

print(f"Train R² = {train_r2:.4f}")
print(f"MSE = {mse:.4f}")
print(f"RMSE = {rmse:.4f}")
print(f"Test R² = {test_r2:.4f}")
```

```python
# Feature importance (earthquake intensity measures)
linear_im_importance = (
    pd.Series(
        baseline.coef_,
        index=feature_names
    )
    .abs()
    .sort_values(ascending=False)
)

print(linear_im_importance)
```

```python
# Plot feature importance for earthquake intensity measures

plt.figure(figsize=(10, 6))

linear_im_importance.plot(kind="barh")

plt.title("Feature Importance of Earthquake Intensity Measures\n(Linear Regression)")
plt.xlabel("Absolute Coefficient Value")
plt.ylabel("Intensity Measure (IM)")

plt.tight_layout()
plt.show()
```

Feature Selection

```python
# feature selection
from sklearn.feature_selection import SequentialFeatureSelector

# new linear regression object
linreg = LinearRegression()

# create feature selection object and fit
feature_sel = SequentialFeatureSelector(linreg, n_features_to_select="auto", cv=5)
feature_sel.fit(X_train_pd, y_train)

# print selected features
selected_features = X.columns[feature_sel.get_support()]
print(f"{len(selected_features)} Stepwise selected features =", selected_features)
```

Reduced Linear Regression

```python
# create dfs with selected features
X_train_fs = X_train_pd[selected_features]
X_test_fs  = X_test_pd[selected_features]

# create new model
linreg_fs = LinearRegression()
linreg_fs.fit(X_train_fs, y_train)
fs_preds = linreg_fs.predict(X_test_fs)

# evaluate
train_r2_fs = linreg_fs.score(X_train_fs, y_train)
mse_fs = mean_squared_error(y_test, fs_preds)
rmse_fs = np.sqrt(mse_fs)
test_r2_fs = r2_score(y_test, fs_preds)

print(f"Train R² = {train_r2_fs:.4f}")
print(f"MSE = {mse_fs:.4f}")
print(f"RMSE = {rmse_fs:.4f}")
print(f"Test R² = {test_r2_fs:.4f}")
```

```python
comparison = pd.DataFrame({
    "Model": ["Full", "Reduced"],
    "Train R²": [train_r2, train_r2_fs],
    "Test R²": [test_r2, test_r2_fs],
    "RMSE": [rmse, rmse_fs]
})

comparison
```

```python
# Feature importance of selected features
importanceFS = (
    pd.Series(linreg_fs.coef_, index=selected_features)
      .abs()
      .sort_values(ascending=False)
)

display(importanceFS.round(2).to_frame("Absolute Coefficient"))

plt.figure(figsize=(8,5))
importanceFS.plot(kind="bar", color="steelblue")
plt.title("Feature Importance (Reduced Linear Regression)")
plt.xlabel("Selected Features")
plt.ylabel("Absolute Coefficient")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Ridge Regression

```python
# Ridge Regression for earthquake intensity measures

from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Ridge regression with cross-validation
ridgeReg = RidgeCV(
    alphas=[0.01, 0.1, 1, 5, 10, 15, 20, 25, 50, 100],
    cv=5
)

# Train using STANDARDIZED earthquake IMs
ridgeReg.fit(X_train_scaled, y_train)

# Predict
RR_y_pred = ridgeReg.predict(X_test_scaled)

# Evaluate
RR_train_r2 = ridgeReg.score(X_train_scaled, y_train)
RR_mse = mean_squared_error(y_test, RR_y_pred)
RR_rmse = np.sqrt(RR_mse)
RR_test_r2 = r2_score(y_test, RR_y_pred)

print(f"Optimal Alpha = {ridgeReg.alpha_}")
print(f"Train R² = {RR_train_r2:.4f}")
print(f"MSE = {RR_mse:.4f}")
print(f"RMSE = {RR_rmse:.4f}")
print(f"Test R² = {RR_test_r2:.4f}")
```

```python
# Ridge Regression Feature Importance

importanceRR = pd.Series(
    ridgeReg.coef_,
    index=X.columns
).abs().sort_values(ascending=False)

print(importanceRR)

plt.figure(figsize=(10, 6))
importanceRR.plot(kind="barh")

plt.title("Ridge Regression - Earthquake IM Importance")
plt.xlabel("Absolute Coefficient Value")
plt.ylabel("Earthquake Intensity Measure")
plt.tight_layout()
plt.show()
```

Lasso

```python
# lasso regression model
from sklearn.linear_model import LassoCV
from sklearn.metrics import mean_squared_error, r2_score

# lassoCV model
lassoReg = LassoCV(alphas=[0.00001, 0.0001, 0.001, 0.005, 0.01, 0.1, 1], cv=5, random_state=42)
lassoReg.fit(X_train_scaled, y_train)
# find optimal alpha
print(f"Optimal Alpha = {lassoReg.alpha_}")
# generate preds
LR_y_pred = lassoReg.predict(X_test_scaled)

# evaluation
LR_train_r2 = lassoReg.score(X_train_scaled, y_train) # how well model fits data
LR_mse = mean_squared_error(y_test, LR_y_pred)
LR_rmse = np.sqrt(LR_mse)
LR_test_r2 = r2_score(y_test, LR_y_pred) # how well model generalizes

print(f"Train R² = {LR_train_r2:.4f}")
print(f"MSE = {LR_mse:.4f}")
print(f"RMSE = {LR_rmse:.4f}")
print(f"R² = {LR_test_r2:.4f}")
```

```python
# Which earthquake IMs were reduced to exactly zero by Lasso?

dropped = X.columns[lassoReg.coef_ == 0]

print("IMs removed by Lasso:")
print(dropped.tolist())
```

```python
#Lasso Regression Feature Importance
importanceLR = pd.Series(lassoReg.coef_, index=feature_names).abs().sort_values()
importanceLR = importanceLR.sort_values(ascending=False)

print(importanceLR)
importanceLR.plot(kind='barh')

plt.title("Lasso Regression - Earthquake IM Importance")
plt.xlabel("Absolute Coefficient Value")
plt.ylabel("Earthquake Intensity Measure")
plt.tight_layout()
plt.show()
```

Elastic Net

```python
from sklearn.linear_model import ElasticNetCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

elastic_cv = ElasticNetCV(
    l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 0.99, 1.0],
    alphas=[0.00001, 0.0001, 0.001, 0.005, 0.01, 0.1, 1, 5, 10, 15, 20, 25, 50, 100],
    cv=5,
    max_iter=10000,
    random_state=42
)

# Train
elastic_cv.fit(X_train_scaled, y_train)

# Predict
en_pred = elastic_cv.predict(X_test_scaled)

# Metrics
rmse = np.sqrt(mean_squared_error(y_test, en_pred))
mae = mean_absolute_error(y_test, en_pred)
r2 = r2_score(y_test, en_pred)

# Rank coefficients by magnitude
coefficients = pd.Series(
    elastic_cv.coef_,
    index=feature_names
).sort_values(key=np.abs, ascending=False)

# Selected IMs
selected_ims = coefficients[coefficients != 0]

print("Optimal alpha:", elastic_cv.alpha_)
print("Optimal L1 ratio:", elastic_cv.l1_ratio_)
print(f"RMSE: {rmse:.4f}")
print(f"MAE: {mae:.4f}")
print(f"R²: {r2:.4f}")

print("\nElastic Net coefficients:")
print(coefficients)

print("\nSelected IMs:")
print(selected_ims.index.tolist())
```

```python
#Elastic Net Regression Feature Importance
importanceEN = pd.Series(elastic_cv.coef_, index=feature_names).abs().sort_values()
importanceEN = importanceEN.sort_values(ascending=False)

print(importanceEN)
importanceEN.plot(kind='barh')

plt.title("Elastic Net Regression - Earthquake IM Importance")
plt.xlabel("Coefficeint Value")
plt.ylabel("Features")
plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
# Random Forest Regression for earthquake intensity measures
```

```python

rf_2 = RandomForestRegressor(random_state=42, n_jobs=-1)

param_grid_2 = {
    'n_estimators': [300, 400, 500],
    'max_depth': [6, 8, 10],
    'min_samples_split': [20, 25, 30, 35],
    'min_samples_leaf': [12, 15, 20],
    'max_features': ['sqrt', 0.2, 0.3],
    'bootstrap': [True, False]
}

grid_search_2 = GridSearchCV(
    estimator=rf_2,
    param_grid=param_grid_2,
    cv=5,
    scoring='r2',
    verbose=1,
    n_jobs=-1
)

grid_search_2.fit(X_train, y_train)

print(f"Best Parameters = {grid_search_2.best_params_}")
print(f"Best CV Score = {grid_search_2.best_score_:.4f}")

# Evaluate
rf_2_pred = grid_search_2.predict(X_test)
rf_2_train_r2 = grid_search_2.score(X_train, y_train)
rf_2_mse = mean_squared_error(y_test, rf_2_pred)
rf_2_rmse = np.sqrt(rf_2_mse)
rf_2_test_r2 = r2_score(y_test, rf_2_pred)

print(f"\nFinal Model Performance:")
print(f"Train R² = {rf_2_train_r2:.4f}")
print(f"MSE = {rf_2_mse:.4f}")
print(f"RMSE = {rf_2_rmse:.4f}")
print(f"Test R² = {rf_2_test_r2:.4f}")
print(f"R² Gap = {rf_2_train_r2 - rf_2_test_r2:.4f}")
```

```python
# Compare Random Forest predictions with actual values

rf_comparison = pd.DataFrame({
    "Actual": y_test,
    "RF_Predicted": rf_2_pred
})

rf_comparison["Error"] = (
    rf_comparison["Actual"] - rf_comparison["RF_Predicted"]
)

print(rf_comparison.head(15))
print("\nActual target range:")
print("Min:", y_test.min())
print("Max:", y_test.max())
print("Mean:", y_test.mean())

print("\nRandom Forest prediction range:")
print("Min:", rf_2_pred.min())
print("Max:", rf_2_pred.max())
print("Mean:", rf_2_pred.mean())
```

```python
# Baseline Random Forest - unrestricted comparison

rf_baseline = RandomForestRegressor(
    n_estimators=500,
    random_state=42,
    n_jobs=-1
)

rf_baseline.fit(X_train, y_train)

rf_base_pred = rf_baseline.predict(X_test)

rf_base_train_r2 = rf_baseline.score(X_train, y_train)
rf_base_test_r2 = r2_score(y_test, rf_base_pred)
rf_base_rmse = np.sqrt(mean_squared_error(y_test, rf_base_pred))

print("Baseline Random Forest:")
print(f"Train R² = {rf_base_train_r2:.4f}")
print(f"Test R² = {rf_base_test_r2:.4f}")
print(f"RMSE = {rf_base_rmse:.4f}")
print(f"R² Gap = {rf_base_train_r2 - rf_base_test_r2:.4f}")
print(f"Prediction Min = {rf_base_pred.min():.4f}")
print(f"Prediction Max = {rf_base_pred.max():.4f}")
```

#XGBoost

```python
#import packages
import xgboost as xgb
import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
# Names of IMs retained by Elastic Net
selected_im_names = selected_ims.index.tolist()

if not selected_im_names:
    raise ValueError("Elastic Net selected no IMs.")

# Use the original, unscaled data for XGBoost
X_train_xgb = X_train[selected_im_names]
X_test_xgb = X_test[selected_im_names]

# Build XGBoost regression model
xgb_model = xgb.XGBRegressor(
    objective="reg:squarederror",
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="rmse"
)

# Train
xgb_model.fit(
    X_train_xgb,
    y_train,
    eval_set=[(X_test_xgb, y_test)],
    verbose=False
)

# Predict
xgb_pred = xgb_model.predict(X_test_xgb)

# Evaluate
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_pred))
xgb_mae = mean_absolute_error(y_test, xgb_pred)
xgb_r2 = r2_score(y_test, xgb_pred)

print("Selected IMs used by XGBoost:")
print(selected_im_names)

print("\nXGBoost performance:")
print(f"RMSE: {xgb_rmse:.4f}")
print(f"MAE: {xgb_mae:.4f}")
print(f"R²: {xgb_r2:.4f}")

# Rank XGBoost feature importance
xgb_importance = pd.Series(
    xgb_model.feature_importances_,
    index=selected_im_names
).sort_values(ascending=False)

print("\nXGBoost feature importance:")
print(xgb_importance)
```

```python
# Check XGBoost for overfitting

xgb_train_pred = xgb_model.predict(X_train_xgb)

xgb_train_r2 = r2_score(y_train, xgb_train_pred)
xgb_test_r2 = r2_score(y_test, xgb_pred)

print(f"Train R² = {xgb_train_r2:.4f}")
print(f"Test R² = {xgb_test_r2:.4f}")
print(f"R² Gap = {xgb_train_r2 - xgb_test_r2:.4f}")
```

```python
# Regularized XGBoost - reduce overfitting

xgb_reg = xgb.XGBRegressor(
    objective="reg:squarederror",
    n_estimators=300,
    learning_rate=0.03,
    max_depth=3,
    min_child_weight=5,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.1,
    reg_lambda=5,
    random_state=42,
    eval_metric="rmse"
)

# Train
xgb_reg.fit(X_train_xgb, y_train)

# Predictions
xgb_reg_train_pred = xgb_reg.predict(X_train_xgb)
xgb_reg_pred = xgb_reg.predict(X_test_xgb)

# Evaluate
xgb_reg_train_r2 = r2_score(y_train, xgb_reg_train_pred)
xgb_reg_test_r2 = r2_score(y_test, xgb_reg_pred)
xgb_reg_rmse = np.sqrt(mean_squared_error(y_test, xgb_reg_pred))
xgb_reg_mae = mean_absolute_error(y_test, xgb_reg_pred)

print("Regularized XGBoost:")
print(f"Train R² = {xgb_reg_train_r2:.4f}")
print(f"Test R² = {xgb_reg_test_r2:.4f}")
print(f"R² Gap = {xgb_reg_train_r2 - xgb_reg_test_r2:.4f}")
print(f"RMSE = {xgb_reg_rmse:.4f}")
print(f"MAE = {xgb_reg_mae:.4f}")
```

```python
# Feature importance - Regularized XGBoost

xgb_reg_importance = pd.Series(
    xgb_reg.feature_importances_,
    index=selected_im_names
).sort_values(ascending=False)

print("Regularized XGBoost Feature Importance:")
print(xgb_reg_importance)
```

```python
#feature plot
# Update feature names to the selected IMs
xgb_reg.get_booster().feature_names = selected_im_names
xgb.plot_importance(
    xgb_reg,
    max_num_features=10,
    height=0.6
)
plt.grid(False)
plt.show()
```

# K-Nearest Neighbors (KNN) Regression

This section develops a K-Nearest Neighbors regression surrogate model to predict the EDP (MIDR) from the synthetic practice dataset.

The purpose of this section is to understand and recreate the KNN surrogate-modeling workflow before applying it to the real IM–EDP database obtained from NLTHA.

## 1. Prepare Data for KNN 

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
```

```python
# KNN Regression with Grid Search

knn = KNeighborsRegressor()

knn_param_grid = {
    "n_neighbors": [3, 5, 7, 9, 11, 15],
    "weights": ["uniform", "distance"],
    "p": [1, 2]
}

knn_grid = GridSearchCV(
    estimator=knn,
    param_grid=knn_param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

knn_grid.fit(X_train_scaled, y_train)

print("Best KNN Parameters:")
print(knn_grid.best_params_)

print(f"Best CV R² = {knn_grid.best_score_:.4f}")
```

```python
# Evaluate Best KNN Model

knn_best = knn_grid.best_estimator_

knn_train_pred = knn_best.predict(X_train_scaled)
knn_test_pred = knn_best.predict(X_test_scaled)

knn_train_r2 = r2_score(y_train, knn_train_pred)
knn_test_r2 = r2_score(y_test, knn_test_pred)

knn_rmse = np.sqrt(
    mean_squared_error(y_test, knn_test_pred)
)

knn_mae = mean_absolute_error(
    y_test,
    knn_test_pred
)

print("KNN Model Performance:")
print(f"Train R² = {knn_train_r2:.4f}")
print(f"Test R² = {knn_test_r2:.4f}")
print(f"R² Gap = {knn_train_r2 - knn_test_r2:.4f}")
print(f"RMSE = {knn_rmse:.4f}")
print(f"MAE = {knn_mae:.4f}")
```

# Gaussian Process Regression (GPR)

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel
```

```python
# Gaussian Process Regression with Grid Search

gpr = GaussianProcessRegressor(
    normalize_y=True,
    random_state=42
)

gpr_param_grid = {
    "kernel": [
        ConstantKernel(1.0) * RBF(length_scale=0.1),
        ConstantKernel(1.0) * RBF(length_scale=0.5),
        ConstantKernel(1.0) * RBF(length_scale=1.0),
        ConstantKernel(1.0) * RBF(length_scale=2.0)
    ],
    "alpha": [1e-4, 1e-3, 1e-2, 1e-1]
}

gpr_grid = GridSearchCV(
    estimator=gpr,
    param_grid=gpr_param_grid,
    cv=5,
    scoring="r2",
    n_jobs=-1
)

gpr_grid.fit(X_train_scaled, y_train)

print("Best GPR Parameters:")
print(gpr_grid.best_params_)

print(f"Best CV R² = {gpr_grid.best_score_:.4f}")
```

```python
# Evaluate Best GPR Model

gpr_best = gpr_grid.best_estimator_

gpr_train_pred = gpr_best.predict(X_train_scaled)
gpr_test_pred = gpr_best.predict(X_test_scaled)

gpr_train_r2 = r2_score(y_train, gpr_train_pred)
gpr_test_r2 = r2_score(y_test, gpr_test_pred)

gpr_rmse = np.sqrt(
    mean_squared_error(y_test, gpr_test_pred)
)

gpr_mae = mean_absolute_error(
    y_test,
    gpr_test_pred
)

print("GPR Model Performance:")
print(f"Train R² = {gpr_train_r2:.4f}")
print(f"Test R² = {gpr_test_r2:.4f}")
print(f"R² Gap = {gpr_train_r2 - gpr_test_r2:.4f}")
print(f"RMSE = {gpr_rmse:.4f}")
print(f"MAE = {gpr_mae:.4f}")
```
